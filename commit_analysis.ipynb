```python
# Vamos usar a API do GitHub para obter diffs dos dois commits mais recentes
def fetch_commit_diff(commit_hash):
    import requests
    repo = "deuspou266-droid/TELEGRAMLEECHv2"
    url = f"https://api.github.com/repos/{repo}/commits/{commit_hash}"
    resp = requests.get(url)
    if resp.status_code != 200:
        print(f"Erro ao buscar commit {commit_hash}: {resp.status_code}")
        return None
    data = resp.json()
    # GitHub API returns patch text under files[].patch
    patches = []
    for f in data.get('files',[]):
        patches.append(f"--- {f.get('filename')}\n" + (f.get('patch') or ""))
    return "\n".join(patches)

# obter lista de commits usando outra chamada
def list_recent_commits(n=5):
    import requests
    repo = "deuspou266-droid/TELEGRAMLEECHv2"
    url = f"https://api.github.com/repos/{repo}/commits"
    resp = requests.get(url, params={"per_page": n})
    if resp.status_code != 200:
        print("Erro ao listar commits", resp.status_code)
        return []
    return [c['sha'] for c in resp.json()]

hashes = list_recent_commits(4)
print("Recent hashes:", hashes)
for h in hashes[:2]:
    diff = fetch_commit_diff(h)
    print(f"--- DIFF for {h} ---\n", diff[:2000])
```
